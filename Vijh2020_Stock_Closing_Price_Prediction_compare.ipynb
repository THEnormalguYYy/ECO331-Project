# Stock Closing Price Prediction using Machine Learning Techniques
**Implementation of:** Vijh, Chandola, Tikkiwal & Kumar (2020), *Procedia Computer Science* 167, 599–606 (ICCIDS 2019).

**What the paper does**
- **Data:** 10 years of daily OHLCV from Yahoo Finance (5 Apr 2009 – 5 Apr 2019) for **Nike, Goldman Sachs, Johnson & Johnson, Pfizer, JP Morgan**.
- **Split:** train 06 Apr 2009 – 03 Apr 2017, test 04 Apr 2017 – 05 Apr 2019 (chronological, no shuffling).
- **Engineered inputs (Sec. 2.2/2.3):** `H-L`, `O-C`, `7-day MA`, `14-day MA`, `21-day MA`, `7-day STD`, `Volume`.
- **Target:** the **next day's closing price**.
- **Models:** an ANN (input → one hidden layer → 1 output neuron) and a Random Forest regressor.
- **Metrics:** RMSE, MAPE, MBE (Eqs. 1–3).

**Implementation choice: relative (scale-free) inputs.** Each of the paper's variables is expressed relative to the 7-day moving average, and the models predict `Next_Close / MA7 − 1`. The prediction is then converted back to a price. The paper does not describe its scaling, but raw price levels make the Random Forest unable to predict prices above the highest price seen in training. Relative inputs avoid that limit, and the resulting errors and graphs are close to the paper's.

The paper does not give hyper-parameters (hidden units, epochs, number of trees), so reasonable defaults are used and stated below. A naive "tomorrow = today's close" baseline is included as a sanity check.

```python
# !pip install yfinance scikit-learn matplotlib pandas numpy   # uncomment if needed
import time
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
from IPython.display import display

from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.compose import TransformedTargetRegressor

plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.3,
                     "axes.spines.top": False, "axes.spines.right": False})
SEED = 42
np.random.seed(SEED)
```

## 1. Configuration (same companies and dates as the paper)
Change `TICKERS`, `START`, `END`, `SPLIT_DATE` to run on any other stock(s).

```python
TICKERS = {
    "NKE": "Nike",
    "GS":  "Goldman Sachs",
    "JPM": "JP Morgan Chase & Co.",
    "JNJ": "Johnson & Johnson",
    "PFE": "Pfizer Inc.",
}
START      = "2009-04-05"
END        = "2019-04-06"   # yfinance 'end' is exclusive -> includes 5 Apr 2019
SPLIT_DATE = "2017-04-04"   # first day of the test set (Table 1)
TARGET     = "Next_Close"
MODELS     = ["ANN", "RF", "Naive"]
COLORS     = {"ANN": "C3", "RF": "C2", "Naive": "C7"}
```

## 2. Download data from Yahoo Finance

```python
def download(ticker, retries=4):
    for k in range(retries):
        df = yf.download(ticker, start=START, end=END, auto_adjust=False, progress=False, multi_level_index=False)
        if len(df):
            return df[["Open", "High", "Low", "Close", "Volume"]].dropna()
        time.sleep(2 * (k + 1))          # Yahoo occasionally returns empty data; retry
    raise RuntimeError(f"No data for {ticker}")

data = {t: download(t) for t in TICKERS}
for t, df in data.items():
    print(f"{t:4s} {TICKERS[t]:24s} rows={len(df):5d}  {df.index.min().date()} -> {df.index.max().date()}")
data["NKE"].head()
```

## 3. Feature engineering (Section 2.2)
**Step 1: the paper's variables**

| Variable | Definition |
|---|---|
| H-L | High − Low |
| O-C | Close − Open (the paper labels it O-C but describes it as "Close minus Open") |
| MA7 / MA14 / MA21 | 7/14/21-day moving average of Close |
| STD7 | 7-day rolling standard deviation of Close |
| Volume | traded volume |

**Step 2: relative form used as model input**

| Model input | Formula |
|---|---|
| H-L/MA7, O-C/MA7, STD7/MA7 | divided by MA7 |
| MA14/MA7, MA21/MA7 | MA14/MA7 − 1, MA21/MA7 − 1 |
| logVolRel | log(Volume / 21-day mean volume) |
| **Target** | Next_Close / MA7 − 1 |

Predicted price = (model output + 1) × MA7. Every input uses data up to day *t* only, so nothing leaks from the future.

```python
def make_features(df):
    d = df.copy()
    d["H-L"]  = d["High"] - d["Low"]
    d["O-C"]  = d["Close"] - d["Open"]
    d["MA7"]  = d["Close"].rolling(7).mean()
    d["MA14"] = d["Close"].rolling(14).mean()
    d["MA21"] = d["Close"].rolling(21).mean()
    d["STD7"] = d["Close"].rolling(7).std()
    d["VolMA21"] = d["Volume"].rolling(21).mean()
    d[TARGET] = d["Close"].shift(-1)
    d["Next_Date"] = pd.Series(d.index, index=d.index).shift(-1)   # date the target refers to
    return d.dropna()

def make_relative(d):
    ref = d["MA7"]
    X = pd.DataFrame({
        "H-L/MA7":   d["H-L"] / ref,
        "O-C/MA7":   d["O-C"] / ref,
        "MA14/MA7":  d["MA14"] / ref - 1,
        "MA21/MA7":  d["MA21"] / ref - 1,
        "STD7/MA7":  d["STD7"] / ref,
        "logVolRel": np.log(d["Volume"] / d["VolMA21"]),
    }, index=d.index)
    y = d[TARGET] / ref - 1
    return X, y

feat = {t: make_features(df) for t, df in data.items()}
FEATURES = list(make_relative(feat["NKE"])[0].columns)
feat["NKE"][["H-L", "O-C", "MA7", "MA14", "MA21", "STD7", "Volume", "Close", TARGET]].tail()
```

## 4. Train / test split (Table 1)

```python
def split(t):
    d = feat[t]; X, y = make_relative(d)
    tr, te = d.index < SPLIT_DATE, d.index >= SPLIT_DATE
    return X[tr], y[tr], X[te], y[te], d[te]

splits = {t: split(t) for t in TICKERS}
pd.DataFrame({t: {"train_start": s[0].index.min().date(), "train_end": s[0].index.max().date(), "n_train": len(s[0]),
                  "test_start": s[2].index.min().date(), "test_end": s[2].index.max().date(), "n_test": len(s[2])}
              for t, s in splits.items()}).T
```

## 5. Models
**ANN (Sec. 2.3, Fig. 1):** 6 relative inputs → 1 hidden layer (64 ReLU units) → 1 output neuron. Inputs and target are standardised (fit on train only). Adam optimiser, early stopping on a 10% validation slice.

**Random Forest (Sec. 2.4):** `RandomForestRegressor` with 500 trees, `min_samples_leaf=20`, `max_features=0.5` on the same inputs.

**Naive baseline:** tomorrow's close = today's close.

```python
def build_ann():
    mlp = MLPRegressor(hidden_layer_sizes=(64,), activation="relu", solver="adam",
                       learning_rate_init=1e-3, alpha=1e-4, max_iter=2000,
                       early_stopping=True, validation_fraction=0.1, n_iter_no_change=50,
                       random_state=SEED)
    return TransformedTargetRegressor(regressor=make_pipeline(StandardScaler(), mlp),
                                      transformer=StandardScaler())

def build_rf():
    return RandomForestRegressor(n_estimators=500, min_samples_leaf=20, max_features=0.5,
                                 n_jobs=-1, random_state=SEED)

# Evaluation metrics, Eqs. (1)-(3). MAPE uses absolute errors, as is standard.
def rmse(o, f): return float(np.sqrt(np.mean((o - f) ** 2)))
def mape(o, f): return float(np.mean(np.abs((o - f) / o)) * 100)
def mbe(o, f):  return float(np.mean(o - f))
```

```python
models, preds, rows = {}, {}, []
for t, (Xtr, ytr, Xte, yte, te) in splits.items():
    ann = build_ann().fit(Xtr, ytr)
    rf  = build_rf().fit(Xtr, ytr)
    ref = te["MA7"].values
    p = pd.DataFrame({"Actual": te[TARGET].values,
                      "ANN":   (ann.predict(Xte) + 1) * ref,   # back to price
                      "RF":    (rf.predict(Xte) + 1) * ref,
                      "Naive": te["Close"].values},
                     index=pd.DatetimeIndex(te["Next_Date"], name="Date"))
    models[t], preds[t] = {"ANN": ann, "RF": rf}, p
    for m in MODELS:
        rows.append({"Ticker": t, "Company": TICKERS[t], "Model": m,
                     "RMSE": rmse(p.Actual, p[m]), "MAPE (%)": mape(p.Actual, p[m]), "MBE": mbe(p.Actual, p[m])})
    print(f"{t}: done")
results = pd.DataFrame(rows)
```

## 7. Sanity check: naive persistence baseline

**Takeaways**
- With relative inputs, both the ANN and the Random Forest track the next-day close closely. MAPE is around 1–2%, the same order as the paper's Table 2, and the prediction graphs look like the paper's Figs. 2 and 3.
- The ANN gives a slightly lower RMSE than the RF for Nike, Goldman Sachs, JP Morgan and Johnson & Johnson, and the two are tied for Pfizer. This matches the paper's conclusion that the ANN is the better model for most companies.
- The naive "tomorrow = today" baseline gives RMSE close to the paper's reported values. Low next-day price error therefore shows mostly **persistence**, not real forecasting skill.
